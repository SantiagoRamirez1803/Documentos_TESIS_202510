### Matriz de adyacencia espacial

```python

# Traer df con los barrios y las coordenadas

import geopandas as gpd
import pandas as pd
from shapely.geometry import Point, Polygon
import matplotlib.pyplot as plt
import ast  # Para convertir cadenas de texto en listas
df_barrios = pd.read_csv('HURTOS POR BARRIO 2022.csv')
print(df_barrios["Coordenadas"].isna().sum())  # Ver cuántas filas están vacías
# Eliminar las filas vacías
df_barrios = df_barrios.dropna(subset=['Coordenadas'])
# Convertir la columna "Coordenadas" de string a lista real
df_barrios["Coordenadas"] = df_barrios["Coordenadas"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else None)

def contar_niveles(coordenadas):
    """ Cuenta cuántos niveles de listas hay en las coordenadas """
    nivel = 0
    while isinstance(coordenadas, list) and len(coordenadas) > 0:
        coordenadas = coordenadas[0]
        nivel += 1
    return nivel

# Aplicar la función a todas las coordenadas
df_barrios["Niveles_Anidamiento"] = df_barrios["Coordenadas"].apply(contar_niveles)

# Ver cuántos niveles diferentes existen
print(df_barrios["Niveles_Anidamiento"].value_counts())
def extraer_coordenadas(coordenadas):
    """ Extrae la lista de coordenadas correcta, eliminando niveles innecesarios de anidación """
    if isinstance(coordenadas, str):
        coordenadas = ast.literal_eval(coordenadas)  # Convertir string a lista si es necesario

    if not isinstance(coordenadas, list) or len(coordenadas) == 0:
        return None  # Si no es lista o está vacía, devolver None

    # Extraer hasta llegar al nivel correcto (eliminando listas anidadas innecesarias)
    while isinstance(coordenadas[0], list) and isinstance(coordenadas[0][0], list):
        coordenadas = coordenadas[0]  # Seguir extrayendo

    return coordenadas  # Devolver la lista de coordenadas corregida

# Aplicar la función para limpiar las coordenadas
df_barrios["Coordenadas_Limpias"] = df_barrios["Coordenadas"].apply(extraer_coordenadas)

# Convertir la columna corregida en polígonos
def convertir_a_poligono(coordenadas):
    """ Convierte listas de coordenadas en polígonos """
    if isinstance(coordenadas, list) and len(coordenadas) > 0:
        try:
            return Polygon(coordenadas)  # Crear el polígono con la lista final de coordenadas
        except Exception as e:
            print(f"Error en coordenadas: {coordenadas}, Error: {e}")
            return None
    return None  # Si no es lista, devolver None

df_barrios["geometry"] = df_barrios["Coordenadas_Limpias"].apply(convertir_a_poligono)

# Verificar si la conversión funcionó
print("Cantidad de barrios sin geometría:", df_barrios["geometry"].isna().sum())
```

```python
# Dejar solo las columnas necesarias
df_barrios = df_barrios[['Barrio_Localidad', 'geometry']]
```

```python
from libpysal.weights import Queen
w = Queen.from_dataframe(df_barrios,idVariable='Barrio_Localidad')
w.transform = 'r'  # para estandarizar las filas
```

### Modelo Poisson Generalizado (Sin espacialidad)

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Definir la variable dependiente y las independientes (todas las demás variables menos la variable dependiente)
formula = 'CANTIDAD_HURTOS ~ Num_Personas + Num_Viviendas + Num_Industria + Num_Comercio + Num_Servicios + Num_Agros + Num_Institucional + Num_Lotes + Num_Parques + Num_MineroEnergetico + Num_Proteccion + Num_Construccion + Num_SinInfo + Distancia_CAI + Num_Colegios +  Num_Cai + Tasa_Hurtos_Lunes + Tasa_Hurtos_Martes + Tasa_Hurtos_Miercoles + Tasa_Hurtos_Jueves + Tasa_Hurtos_Viernes + Tasa_Hurtos_Sabado + Tasa_Hurtos_Domingo'
modelo_poisson = smf.glm(formula=formula, data=data_mod, family=sm.families.Poisson()).fit()

print(modelo_poisson.summary())
# ¿Qué variables son significativas?
modelo_poisson.pvalues[modelo_poisson.pvalues < 0.05].index.tolist()
```

### We use “moran.test” in R (Cliff Ord 1981), which performs a hypothesis test with a null hypothesis of independence between variables. The p-values for the three crime types are significantly less than 0.01, indicating that the counts for each crime do exhibit spatial dependency.

```python
# Asegúrate de importar lo necesario
import geopandas as gpd
from esda.moran import Moran
from splot.esda import moran_scatterplot

# Unir para tener geometría + hurtos en un solo GeoDataFrame
gdf_moran = df_barrios.merge(data_mod[['Barrio_Localidad', 'CANTIDAD_HURTOS']], on='Barrio_Localidad')

# Asegúrate que sea GeoDataFrame
gdf_moran = gpd.GeoDataFrame(gdf_moran, geometry='geometry', crs='EPSG:4326')
# Aplicar test de Moran
moran = Moran(gdf_moran['CANTIDAD_HURTOS'], w)

# Resultados
print("Moran’s I:", moran.I)
print("p-value:", moran.p_sim)
# Graficar el scatterplot
fig, ax = moran_scatterplot(moran)
ax.set_title("Moran Scatterplot - Hurtos por Barrio en Bogotá")
plt.show()
```

#### El valor positivo de Moran’s I indica agrupamiento espacial de valores similares (es decir, valores altos están cerca de otros altos, o bajos con bajos). El p-valor es muy bajo (0.002), por lo tanto, el resultado es estadísticamente significativo. Entonces, podemos concluir que hay una autocorrelación espacial positiva significativa en los datos.
